### Définition système de Rossler

#### Paramètres

```python
t_span = (0, 100)
n_t_steps = 10000
```

```python
sim = RosslerSimulation(
    a=0.2,
    b=0.2,
    c=5.7,
    t_span=t_span,
    n_t_steps=n_t_steps,
    X0=[1.0, 1.0, 1.0]
)

sol = sim.run()
sim.plot_solution()
sim.save("rossler-results.npz")
```

```python
params = {"a": 0.2, "b": 0.2,"c": 5.7}
```

```python
sim2 = RosslerSimulation.from_parameters(
    params=params,
    t_span=t_span,
    n_t_steps=n_t_steps,
    X0=[1.0, 1.0, 1.0]
)
```

```python
sim2.run()
sim2.plot_solution()
```
