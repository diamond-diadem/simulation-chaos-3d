### Définition système de Halvorsen

#### Paramètres

```python
# Define the time span for the simulation: from t=0 to t=100
t_span = (0, 100)

# Set the number of time steps for the simulation
n_t_steps = 10000
```

```python
# Create a HalvorsenSimulation instance with specified parameters
sim = HalvorsenSimulation(
    a = 1.4,
    t_span=t_span,        # Time span for simulation (from cell 4)
    n_t_steps=n_t_steps,  # Number of time steps (from cell 4)
    X0=[1.0, 0.0, 0.0]    # Initial condition
)

sol = sim.run()  # Run the simulation

sim.plot_solution(backend="plotly")  # Plot the solution using Plotly

sim.save("halvorsen-results.npz")  # Save the results to a file
```
