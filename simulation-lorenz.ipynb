## Simulation de Lorenz

```python
from machinery import LorenzSimulation  # Import the LorenzSimulation class from the machinery module
```

### Définition système de Lorenz

#### Paramètres

```python
# Define the time span for the simulation: from t=0 to t=100
t_span = (0, 100)

# Set the number of time steps for the simulation
n_t_steps = 10000
```

```python
# Create a LorenzSimulation instance with specified parameters
sim = LorenzSimulation(
    sigma=10.0,           # Prandtl number
    rho=28.0,             # Rayleigh number
    beta=8.0 / 3.0,       # Geometric factor
    t_span=t_span,        # Time span for simulation (from cell 4)
    n_t_steps=n_t_steps,  # Number of time steps (from cell 4)
    X0=[1.0, 1.0, 1.0]    # Initial condition
)

sol = sim.run()  # Run the simulation

sim.plot_solution(backend="plotly")  # Plot the solution using Plotly

sim.save("lorenz-results.npz")  # Save the results to a file
```

```python
import json

# Load configuration from a JSON file
with open("config-lorenz.json") as f:
    cfg = json.load(f)

# Create a LorenzSimulation instance using parameters from the config
sim = LorenzSimulation.from_parameters(cfg["params"], **cfg["meta"])

# Run the simulation
sim.run()

# Plot the solution using Plotly
sim.plot_solution(backend="plotly")
```

```python
import yaml

# Load configuration from a YAML file
with open("config-lorenz.yml") as f:
    cfg = yaml.safe_load(f)

# Create a LorenzSimulation instance using parameters from the YAML config
sim = LorenzSimulation.from_parameters(cfg["params"], **cfg["meta"])

# Run the simulation
sim.run()

# Plot the solution using Plotly
sim.plot_solution(backend="plotly")
```
