## Simulation de Lorenz

```python
from machinery import LorenzSimulation
```

### Définition système de Lorenz

#### Paramètres

```python
t_span = (0, 100)
n_t_steps = 10000
```

```python
sim = LorenzSimulation(
    sigma=10.0,
    rho=28.0,
    beta=8.0 / 3.0,
    t_span=t_span,
    n_t_steps=n_t_steps,
    X0=[1.0, 1.0, 1.0]
)

sol = sim.run()
sim.plot_solution(backend="plotly")
sim.save("lorenz-results.npz")
```

```python
import json
with open("config-lorenz.json") as f:
    cfg = json.load(f)

sim = LorenzSimulation.from_parameters(cfg["params"], **cfg["meta"])
sim.run()
sim.plot_solution(backend="plotly")
```

```python
import yaml
with open("config-lorenz.yml") as f:
    cfg = yaml.safe_load(f)

sim = LorenzSimulation.from_parameters(cfg["params"], **cfg["meta"])
sim.run()
sim.plot_solution(backend="plotly")
```
