## Simulation de Lorenz

```python
import numpy as np

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

from scipy.integrate import solve_ivp

from machinery import LorenzSystem
```

### Définition système de Lorenz

#### Paramètres

```python
sigma = 10.0
rho = 28.0
beta = 8.0 / 3.0
```

#### Initialisation fonction

```python
f = LorenzSystem(sigma=sigma, rho=rho, beta=beta)
```

#### Intervalle de temps

```python
t_span = (0.0, 40.0)
t_eval = np.linspace(*t_span, 10000)
```

#### Conditions initiales

```python
x0 = 1.0
y0 = 1.0
z0 = 1.0

X0 = np.array([x0, y0, z0])
```

### Résolution

```python
sol = solve_ivp(f, t_span, X0, t_eval=t_eval, method='RK45')
```
