#### Paramètres

```python
t_span = (0, 100)
t_eval = np.linspace(*t_span, 10000)
```

```python
sim = LorenzSimulation(
    sigma=10.0,
    rho=28.0,
    beta=8.0 / 3.0,
    t_span=t_span,
    t_eval=t_eval,
    X0=[1.0, 1.0, 1.0]
)

sol = sim.run()
sim.plot_solution()
sim.save("lorenz.npz")
```
